```python
# YOLOv8 설치
!pip install ultralytics
# 필요한 라이브러리 임포트
import os
from ultralytics import YOLO
import shutil
from google.colab import drive
from google.colab import files
# 파일 선택기를 통한 업로드
uploaded = files.upload()
# 코랩 셀에서 ! 를 앞에 붙여서 실행
!mkdir -p /content/dataset
# Unzip the uploaded file into the dataset directory
# Use the actual filename from the uploaded dict
uploaded_filename = list(uploaded.keys())[0]
!unzip -o "/content/{uploaded_filename}" -d /content/dataset

# --- Added steps to verify and fix the directory structure ---
# Verify the contents of the dataset directory after unzipping
print("\nContents of /content/dataset/ after unzipping:")
!ls -R /content/dataset/

# Explicitly create the train/images directory if it doesn't exist
!mkdir -p /content/dataset/train/images

# Assuming the training images are in a folder named 'train/images' within the zip,
# but if not, you might need to adjust the source path below
# This copies any files found in the dataset's top-level 'train/images' to the target
# If the unzip created the directory correctly, this just confirms.
# If the structure is different, you might need to manually copy from the correct path.
# Example: if the zip extracted to /content/dataset/Test.v1i.yolov8/train/images, you'd use that path below.
# Let's assume the standard structure for now.
if os.path.exists('/content/dataset/train/images') and not os.listdir('/content/dataset/train/images'):
    print("\nWarning: /content/dataset/train/images is empty. Check your zip file structure.")
    # You might need to add specific copying logic here if the zip structure is non-standard.
    # For example: !cp /content/dataset/some_other_folder/train/images/* /content/dataset/train/images/

# test/images 폴더 생성 및 복사 (기존 코드 유지)
!mkdir -p /content/dataset/test/images
!cp /content/dataset/train/images/* /content/dataset/test/images/
# valid/images 폴더 생성 및 복사 (기존 코드 유지)
!mkdir -p /content/dataset/valid/images
!cp /content/dataset/train/images/* /content/dataset/valid/images/
# valid/labels 폴더 생성 및 복사 (기존 코드 유지)
!mkdir -p /content/dataset/valid/labels
!cp /content/dataset/train/labels/* /content/dataset/valid/labels/
# --- End of added steps ---

# 학습 시작
from ultralytics import YOLO

# Segmentation 모델로 학습
model = YOLO('yolov8l-seg.pt')  # segmentation 모델 사용

results = model.train(
    data='/content/dataset/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    name='yolov8_seg_custom',
    project='/content/runs/train',
    patience=10,
    save=True,
    device=0
)
# 학습된 모델 로드(폴더에 맞도록)
model = YOLO('/content/runs/train/yolov8_seg_custom2/weights/best.pt')

# 테스트 실행
results = model.predict(
    source='/content/dataset/test/images',  # 테스트 이미지 경로
    save=True,                              # 결과 저장
    save_txt=True,                          # 텍스트 결과도 저장
    conf=0.5,                               # 신뢰도 임계값
    project='/content/runs/detect',         # 저장 프로젝트 경로
    name='test_results'                     # 폴더 이름 지정
)
import cv2
import numpy as np # numpy 임포트 추가

def draw_enhanced_results(image_path, results):
    """세그멘테이션 마스크를 이미지 위에 시각화"""
    # 이미지 로드
    img = cv2.imread(image_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # 클래스 색상 정의
    class_names = ['dotted-line', 'left_lane', 'left_line', 'right_lane', 'right_line']
    colors = [(255, 0, 0), (0, 255, 0), (0, 0, 255), (255, 255, 0), (255, 0, 255)]

    for result in results:
        masks = result.masks  # 마스크 객체

        if masks is not None and masks.data is not None:
            mask_data = masks.data.cpu().numpy()  # (N, H, W)
            cls_list = result.boxes.cls.cpu().numpy().astype(int)  # 클래스 인덱스 리스트
            conf_list = result.boxes.conf.cpu().numpy()  # 신뢰도

            for i in range(len(mask_data)):
                mask = mask_data[i]
                cls_id = cls_list[i]
                conf = conf_list[i]

                # 마스크 색상 및 클래스 이름
                color = colors[cls_id % len(colors)]
                class_name = class_names[cls_id] if cls_id < len(class_names) else f'class_{cls_id}'
                label = f'{class_name}: {conf:.2f}'

                # 마스크를 컬러로 적용
                colored_mask = np.zeros_like(img_rgb, dtype=np.uint8)
                for c in range(3):
                    colored_mask[:, :, c] = mask * color[c]

                # 반투명하게 overlay
                img_rgb = cv2.addWeighted(img_rgb, 1.0, colored_mask, 0.4, 0)

                # 마스크 경계 추출하여 텍스트 위치 추정
                contours, _ = cv2.findContours((mask * 255).astype(np.uint8), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
                if contours:
                    x, y, w, h = cv2.boundingRect(contours[0])
                    # 라벨 텍스트
                    (text_width, text_height), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.7, 2)
                    cv2.rectangle(img_rgb, (x, y - text_height - 10), (x + text_width, y), color, -1)
                    cv2.putText(img_rgb, label, (x, y - 5), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)

    return img_rgb
import matplotlib.pyplot as plt

# Assuming test_images is defined before this loop.
# It's defined later in the original code in the video creation section.
# Let's define it here for clarity for the visualization loop.
image_dir_test = '/content/dataset/test/images/'
if os.path.exists(image_dir_test):
    test_images = sorted([img for img in os.listdir(image_dir_test) if img.lower().endswith(('.png', '.jpg', '.jpeg'))])
else:
    test_images = []
    print(f"Warning: Test image directory not found at {image_dir_test}")


for i, img_name in enumerate(test_images[:5]):  # 처음 5개만
    img_path = f'{image_dir_test}/{img_name}' # Use the defined test_images directory path

    if not os.path.exists(img_path):
        print(f"Skipping {img_path} as it does not exist.")
        continue

    # 예측 실행
    result = model.predict(img_path, conf=0.5, verbose=False)

    # 향상된 시각화
    enhanced_img = draw_enhanced_results(img_path, result)

    # 결과 표시
    plt.figure(figsize=(12, 8))
    plt.imshow(enhanced_img)
    plt.title(f'Detection Result: {img_name}')
    plt.axis('off')
    plt.show()
import shutil
from google.colab import files

# 1. zip 파일로 압축 (기존 zip 파일이 있다면 덮어씌움)
shutil.make_archive('/content/runs_backup', 'zip', '/content/runs')

# 2. 다운로드 링크 제공
files.download('/content/runs_backup.zip')
#동영상으로 만들기

import os
import cv2
import numpy as np # numpy 임포트 추가
import matplotlib.pyplot as plt

# YOLO 예측 + 시각화 함수 (콜라보용)
image_dir = '/content/dataset/test/images/'
# Redefine test_images here again, as the previous definition was local to the plot loop block
test_images = sorted([img for img in os.listdir(image_dir) if img.lower().endswith(('.png', '.jpg', '.jpeg'))])


frames = []

for i, img_name in enumerate(test_images[:5]):  # 처음 5개만
    img_path = os.path.join(image_dir, img_name)

    if not os.path.exists(img_path):
        print(f"Skipping {img_path} for video creation as it does not exist.")
        continue


    # 예측 실행
    result = model.predict(img_path, conf=0.5, verbose=False)

    # 향상된 시각화
    enhanced_img = draw_enhanced_results(img_path, result)

    # RGB -> BGR (OpenCV용)
    # Ensure enhanced_img is a numpy array before converting color
    if not isinstance(enhanced_img, np.ndarray):
        enhanced_img = np.array(enhanced_img)

    bgr_img = cv2.cvtColor(enhanced_img, cv2.COLOR_RGB2BGR)

    # 모든 프레임 사이즈를 동일하게 유지
    bgr_img = cv2.resize(bgr_img, (1280, 720))

    frames.append(bgr_img)

# 🔄 동영상 생성
video_path = "/content/detection_results.mp4"
fps = 2  # 0.5초 간격 = 2프레임/초
if frames: # Only proceed if there are frames
    height, width, _ = frames[0].shape
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # 코덱 설정

    out = cv2.VideoWriter(video_path, fourcc, fps, (width, height))

    for frame in frames:
        out.write(frame)

    out.release()
    print(f"✅ 영상 저장 완료: {video_path}")

    # Colab에서 다운로드 링크 제공
    from google.colab import files
    files.download(video_path)
else:
    print("No frames were processed for video creation.")
```
